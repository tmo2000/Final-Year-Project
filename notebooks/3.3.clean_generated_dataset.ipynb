```python
import pandas as pd
```

```python
cleaned_df = pd.read_csv("C:\\Users\\temisola\\Final-Year-Project\\datasets\\full_dataset_checkpoint.csv")

print(cleaned_df.shape)
cleaned_df.head()
```

```python
translation_cols = [c for c in cleaned_df.columns if "_translation" in c]
interpretation_cols = [c for c in cleaned_df.columns if "_interpretation" in c]
```

```python
required_cols = translation_cols + interpretation_cols

cleaned_df = cleaned_df[
    cleaned_df[required_cols]
    .apply(lambda x: x.astype(str).str.strip() != "")
    .all(axis=1)
]
```

```python
base_columns = [
    "entry_id",
    "yoruba",
    "english_literal",
    "interpretation_gold"
]
```

```python
keep_columns = base_columns + translation_cols + interpretation_cols

cleaned_df = cleaned_df[keep_columns]
```

```python
print(cleaned_df.shape)
cleaned_df.head()
```

```python
before_rows = len(cleaned_df)

ai_required_cols = [
    "deepseek_translation", "deepseek_interpretation",
    "claude_translation", "claude_interpretation",
    "gemini_translation", "gemini_interpretation",
    "llama_translation", "llama_interpretation",
    "mistral_translation", "mistral_interpretation"
]

mask_complete = (
    cleaned_df[ai_required_cols].notna().all(axis=1) &
    cleaned_df[ai_required_cols].apply(
        lambda col: col.astype(str).str.strip() != ""
    ).all(axis=1)
)

cleaned_df = cleaned_df[mask_complete].copy()

after_rows = len(cleaned_df)

print(f"Rows before cleaning: {before_rows}")
print(f"Rows after cleaning:  {after_rows}")
print(f"Rows removed:         {before_rows - after_rows}")
```

```python
cleaned_df.to_csv( "C:\\Users\\temisola\\Final-Year-Project\\datasets\\yoruba_proverb_final_cleaned.csv", index=False, encoding="utf-8-sig")
```
